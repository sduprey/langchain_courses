```python
import sys; sys.path.append('..')  # so `utils` and `eval` import from the repo root
from utils import ask
```

A prompt function: text in, single label out. Keep the output tight so it's parseable.

```python
LABELS = ["positive", "negative", "neutral"]

FEWSHOT = """Classify the sentiment as positive, negative, or neutral.
Reply with ONE lowercase word and nothing else.

Text: "I love this, works perfectly" -> positive
Text: "Absolute garbage, crashed twice" -> negative
Text: "It is a phone" -> neutral
"""

def classify(text):
    out = ask(FEWSHOT + f'Text: "{text}" -> ').strip().lower()
    # keep only a known label if the model rambles
    for lab in LABELS:
        if lab in out:
            return lab
    return out

print(classify("The screen broke after one day"))
```
