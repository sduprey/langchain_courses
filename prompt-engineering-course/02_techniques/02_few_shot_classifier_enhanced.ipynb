# Build a few-shot classifier

**Session 2 · small model (`llama3.2:3b`)**

A real classifier you'll evaluate and iterate on in Session 3. Prompt engineering with a
measurable goal: text in, one label out, tight enough to parse.

```python
import sys; sys.path.append('..')  # so `utils` and `eval` import from the repo root
from utils import ask, SMALL_MODEL
```

A prompt function: text in, single label out. Keep the output tight so it's parseable.

```python
LABELS = ["positive", "negative", "neutral"]

FEWSHOT = """Classify the sentiment of a product review as positive, negative, or neutral.
Reply with ONE lowercase word and nothing else.

Text: "I love this, works perfectly" -> positive
Text: "Absolute garbage, crashed twice" -> negative
Text: "It runs on two AA batteries" -> neutral
"""

def classify(text):
    out = ask(FEWSHOT + f'Text: "{text}" -> ', model=SMALL_MODEL).strip().lower()
    # keep only a known label if the model rambles
    for lab in LABELS:
        if lab in out:
            return lab
    return out

for t in ["The screen broke after one day", "Ten out of ten, would not recommend"]:
    print(f"{classify(t):9} | {t}")
```
