### Vague vs specific

```python
vague = "Summarize this."
text = "(paste a paragraph here)"
print(ask(f"{vague}\n\n{text}"))
```

```python
specific = chat([
  {"role":"system","content":"You are an editor. Be concise."},
  {"role":"user","content": f"Summarize the text in 3 bullets, max 60 words, for a non-technical manager. If unclear, ask one question.\n\nTEXT:\n\"\"\"{text}\"\"\""}
])
print(specific)
```
