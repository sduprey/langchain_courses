### Worked example

Measure how many *distinct* answers you get at each temperature. Swap the prompt for a creative vs a factual task and compare.


```python
# Worked example: quantify output variability at each temperature
def variability(prompt, temperature, n=5):
    outs = [ask(prompt, temperature=temperature).strip() for _ in range(n)]
    return outs, len(set(outs))

for temp in (0, 0.7, 1.0):
    outs, distinct = variability("Give me a one-line startup idea for a bakery.", temp)
    print(f"temp={temp}: {distinct}/{len(outs)} distinct")
    for o in outs:
        print("   ", o[:70])
```
