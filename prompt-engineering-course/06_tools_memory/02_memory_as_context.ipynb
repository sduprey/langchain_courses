### Worked example

A tiny curated memory store: `remember()` keeps only durable facts, `reply()` re-injects them into the system prompt on later turns.


```python
# Worked example: persist a fact, curate it, re-inject it
MEMORY = []   # curated durable facts about the user

def remember(fact):
    durable = any(k in fact.lower() for k in
                  ["prefer", "name is", "allerg", "timezone", "always", "never"])
    if durable and fact not in MEMORY:
        MEMORY.append(fact)
    return durable

def reply(user_msg):
    system = "You are a helpful assistant."
    if MEMORY:
        system += " Known facts about the user:\n- " + "\n- ".join(MEMORY)
    return chat([{"role": "system", "content": system},
                 {"role": "user", "content": user_msg}])

remember("The user prefers answers in metric units.")   # durable -> kept
remember("The user said hi.")                            # chatter  -> dropped
print("MEMORY:", MEMORY)
print(reply("How far is it from London to Paris?"))      # should answer in km
```
