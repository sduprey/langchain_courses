### Worked example

The same fact at the start, middle, and end of a long context, plus a run with extra irrelevant chunks. Ask the identical question each time.


```python
# Worked example: position matters, and noise hurts
FACT   = "The launch code is TIGER-7."
FILLER = ("The team met Tuesday to review the roadmap. The cafeteria now serves oat milk. "
          "Parking is being repaved next month. ") * 8
Q = "What is the launch code? Answer with the code only."

def context(position):
    half = len(FILLER) // 2
    return {
        "start":  FACT + " " + FILLER,
        "middle": FILLER[:half] + " " + FACT + " " + FILLER[half:],
        "end":    FILLER + " " + FACT,
    }[position]

for pos in ("start", "middle", "end"):
    out = ask(f"Use only this context:\n{context(pos)}\n\n{Q}")
    print(f"  {pos:6}: {out.strip()[:40]}")

noisy = context("middle") + (" The quarterly figures were strong. " * 20)
print(f"  middle+noise: {ask(f'Use only this context:\n{noisy}\n\n{Q}').strip()[:40]}")
```
