### Worked example

A verbose prompt vs a trimmed one: tokens, latency, and projected monthly cost side by side.


```python
# Worked example: trim the prompt, measure the saving, price it
import time
from utils import count_tokens

VERBOSE = ("I would be very grateful if you could kindly take a moment to explain, in a "
           "clear and accessible way, the concept of retrieval-augmented generation, "
           "ideally in around two sentences. Thank you so much.")
TIGHT = "Explain retrieval-augmented generation in 2 sentences."

PRICE_IN, PRICE_OUT = 0.15 / 1_000_000, 0.60 / 1_000_000   # $ per token
CALLS_PER_DAY = 100_000

for name, p in [("verbose", VERBOSE), ("tight", TIGHT)]:
    t0 = time.time(); out = ask(p); dt = time.time() - t0
    tin, tout = count_tokens(p), count_tokens(out)
    monthly = (tin * PRICE_IN + tout * PRICE_OUT) * CALLS_PER_DAY * 30
    print(f"  {name:8} in={tin:3} out={tout:3}  {dt:4.1f}s  ~${monthly:,.0f}/month")
```
